```python
def coloring(g):
    c = {}
    for v in g:
        used = {c[x] for x in g[v] if x in c}
        c[v] = 1
        while c[v] in used:
            c[v] += 1
    return c


n = int(input("Enter number of territories: "))
e = int(input("Enter number of adjacent pairs: "))
p = input("Enter adjacent pairs: ").upper().split()

g = {}
for x in p:
    a, b = x
    g.setdefault(a, []).append(b)
    g.setdefault(b, []).append(a)

print("\n===== GRAPH =====")

# Find root
root = next(v for v in g if len(g[v]) > 1)

# Print root
print("        " + root)

children = g[root]

if len(children) == 2:
    a, b = children

    if b in g[a]:
        print("       / \\")
        print("      " + a + "---" + b)
    else:
        print("       / \\")
        print("      " + a + "   " + b)

    # Remaining connections
    for child in children:
        for x in g[child]:
            if x != root and x not in children:
                print("      |")
                print("      " + x)

else:
    for child in children:
        print("        |")
        print("        " + child)


print("\n===== COLORS =====")
c = coloring(g)

for v in c:
    print(v, "-> Color", c[v])
```
