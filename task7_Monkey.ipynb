```python
class MonkeyBanana:
    def __init__(self):
        self.state = {
            'on_floor(Monkey)',
            'at(Monkey, Window)',
            'at(Box, Window)'
        }

        self.operators = {
            'Push Box': {
                'preconds': {'on_floor(Monkey)', 'at(Monkey, Window)', 'at(Box, Window)'},
                'add': {'at(Monkey, Center)', 'at(Box, Center)'},
                'del': {'at(Monkey, Window)', 'at(Box, Window)'}
            },
            'Climb Box': {
                'preconds': {'on_floor(Monkey)', 'at(Monkey, Center)', 'at(Box, Center)'},
                'add': {'on_box(Monkey)'},
                'del': {'on_floor(Monkey)'}
            },
            'Grasp Banana': {
                'preconds': {'on_box(Monkey)', 'at(Box, Center)', 'at(Monkey, Center)'},
                'add': {'has(Monkey, Banana)'},
                'del': set()
            }
        }

    def apply(self, action):
        op = self.operators[action]
        if op['preconds'] <= self.state:
            self.state -= op['del']
            self.state |= op['add']
            print("\nAction:", action)
            print("Preconditions satisfied")
            print("New state:", self.state)

    def solve(self):
        print("===== MONKEY BANANA PROBLEM =====")
        print("\nInitial State:")
        print("Monkey is at Window")
        print("Box is at Window")
        print("Monkey is on Floor")
        print("Banana is at Center")

        for action in ['Push Box', 'Climb Box', 'Grasp Banana']:
            self.apply(action)

        print("\n===== FINAL RESULT =====")
        print("SUCCESS!")
        print("Monkey has reached the banana.")


MonkeyBanana().solve()
```
